# 00 - Set up the Olist project

Create the `olist_` schemas and a managed landing volume in an existing catalog.
Import this standalone notebook, select **Serverless** and choose **Run all**.
Keep the default catalog unless the project uses another existing catalog.
The notebook is self-contained and requires no shared-notebook imports.


```python
import re

try:
    catalog = dbutils.widgets.get("catalog").strip()
except Exception:
    dbutils.widgets.text("catalog", spark.sql("SELECT current_catalog()").first()[0], "Existing catalog")
    catalog = dbutils.widgets.get("catalog").strip()

if not re.fullmatch(r"[A-Za-z_][A-Za-z0-9_]*", catalog):
    raise ValueError("Use an existing catalog with letters, numbers and underscores.")

for layer in ["landing", "bronze", "silver", "quality"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.`olist_{layer}`")
spark.sql(f"CREATE VOLUME IF NOT EXISTS `{catalog}`.`olist_landing`.`raw`")

print("Setup complete.")
print("Catalog:", catalog)
print("CSV destination:", f"/Volumes/{catalog}/olist_landing/raw")
print("Next step: upload the five original Olist CSVs to this volume.")
```

## Expected result

The output prints `Setup complete` and the landing-volume path.
Upload the five original Olist CSVs directly to that volume before running Bronze.
This notebook creates project resources; it does not ingest transaction data.
